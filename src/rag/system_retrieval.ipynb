```python
from typing import List

def preprocess(s: str) -> List[str]:

    def to_lowercase(s: str) -> str:
        return s.lower()
    
    def tokenize(s: str) -> List[str]:
        return s.split(' ')
    
    def filtering(ls: List[str]) -> List[str]:
        r = [''.join(filter(str.isalnum, s)) for s in ls]
        return list(filter(str.isalnum, r))
    
    return filtering(tokenize(to_lowercase(s)))
```

```python
preprocess("Machine Learning é fantástico!")
```

```python
preprocess("Artificial Intelligence & Machine Learning")
```
